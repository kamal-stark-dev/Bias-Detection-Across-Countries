````python
import pandas as pd
import re
````

````python
df = pd.read_csv(r"../raw/1000-US-Ind-Aus.csv")
df.head()
````

````python
# drop duplicates and nulls

df.drop_duplicates(subset="headline", inplace=True)
df.dropna(subset=["headline", "country"], inplace=True)
````

````python
# normalize text

def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-z0-9\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["headline_clean"] = df["headline"].apply(clean_text)
````

````python
print(df[["headline", "headline_clean"]].head(7))
````

````python
# add id
df.insert(0, "id", range(1, len(df) + 1))
df.head()
````

````python
df.to_csv("headlines_standardized.csv", index=False)
print("✅ Dataset standardized and saved.")
````
