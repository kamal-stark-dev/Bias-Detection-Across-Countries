### STAGE 2 — Country-wise Bias Analysis

🧠 Questions we will answer

1. Which country has higher average bias?

2. Which country uses more emotional language?

3. Which country relies more on framing?

4. Are differences statistically meaningful?

>"Rather than relying on binary bias labels, we propose a continuous Bias Index constructed from linguistic indicators including subjectivity, emotional intensity, and framing language. This enables fine-grained cross-country comparison of headline bias patterns.”

```python
import pandas as pd
```

```python
df = pd.read_csv("headlines_with_bias_index.csv")
df.head(2)
```

```python
summary = df.groupby("country").agg({
    "bias_index": ["mean", "std"],
    "subjectivity": "mean",
    "emotional_intensity": "mean",
    "framing_score": "mean",
    "loaded_binary": "mean"
})

summary.columns = [
    "bias_mean",
    "bias_std",
    "subjectivity_mean",
    "emotion_mean",
    "framing_mean",
    "loaded_language_mean"
]

summary = summary.reset_index()
```

```python
summary.to_csv("country_bias_summary.csv", index=False)
summary
```
